# Run the bot

```python
from src.llm import load_model
from src.rag import build_retriever

model, processor = load_model()
retriever = build_retriever()
```

```python
from src.chatbot import answer_question, add_to_history, format_sources

history = []
while True:
    question = input("You: ").strip()
    if question.lower() in ["quit", "exit", "bye"]:
        break
    if not question:
        continue
    result = answer_question(question, model, processor, retriever, history)
    print("\nChatbot:", result["answer"])
    print("\nSources:\n" + format_sources(result["sources"]) + "\n")
    add_to_history(history, question, result["answer"])
```
